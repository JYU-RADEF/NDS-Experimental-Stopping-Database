````python
df_harmonized = harmonize_energy_units(df_original, to="MeV/u")
df_harmonized
````

````python
projectiles = list(df_original["projectile_name"].unique())
targets = list(df_original["target_name"].unique())

print(f"number of projectiles: {len(projectiles)}")
print(f"number of targets: {len(targets)}")

unique_combinations = df_original[["projectile_name", "target_name"]].drop_duplicates()
print(f"number of unique combinations: {len(unique_combinations)}")
unique_combinations
````

````python
from ipywidgets import interact

df_elemental_only = df_harmonized[
    df_harmonized["target_name"].apply(is_element_in_periodic_table)
]

df_elemental_only
````

````python
df = df_elemental_only.sort_values(["projectile_name"])
projectiles = list(df_elemental_only["projectile_name"].unique())


def filter_by_ion(ion: str):
    df_ = df.copy()

    df_ = df_[df_["projectile_name"] == ion]

    fig = px.scatter(
        df_,
        x="energy",
        y="stopping_power",
        symbol="stopping_unit",
        color="target_name",
        log_x=True,
        log_y=True,
        hover_data=["energy", "energy_unit", "stopping_unit"],
        width=1200,
        height=600,
    )
    # fig.show()
    return fig


f = interact(filter_by_ion, ion=projectiles)
````

````python
df = df.sort_values(["target_name"])
targets = df["target_name"].unique()


def filter_by_target(target: str):
    df_ = df.copy()

    df_ = df_[df_["target_name"] == target]

    fig = px.scatter(
        df_,
        x="energy",
        y="stopping_power",
        color="stopping_unit",
        symbol="projectile_name",
        log_x=True,
        log_y=True,
        hover_data=["energy", "energy_unit", "stopping_unit"],
        width=1200,
        height=600,
    )

    fig.show()


selected_ion = interact(filter_by_target, target=targets)
````
